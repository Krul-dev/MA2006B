# Exponenciación Binaria

En este notebook, mostraremos de manera práctica la diferencia en la velocidad de ejecución entre el algoritmo de exponenciación binaria comparado con el algoritmo de exponenciación ingenua, sin optimizar. Empezaremos por cargar las funciones que necesitamos:

```python
import sys
import importlib.util

if 'google.colab' in sys.modules:
    runtime = "Google Colab"
    if importlib.util.find_spec("cryptocalc") is None:
        print("   Installing MA2006B from GitHub\n")
        !pip install git+https://github.com/Krul-dev/MA2006B.git
else:
    runtime = "Local environment"

import cryptocalc

print(f"\n========= Notebook execution context =========")
print(f"              Runtime: {runtime}")
print(f"       Python version: {sys.version.split()[0]}")
print(f"   CryptoCalc version: {cryptocalc.__version__}\n")

from cryptocalc import (
    RFC_3526_SAFE_PRIME,
    exp_mod,
    naive_exp_mod
)
```

Ahora fijaremos un número primo que utilizaremos como módulo y un valor base.

```python
p = RFC_3526_SAFE_PRIME
g = 2
x = (g, p)
x
```

Para comparar los dos algoritmos, mediremos el tiempo que llevar elevar el punto base a la potencia $n = 100$ usando ambos métodos. Con exponenciación binaria:

```python
%time exp_mod(x, 100)
```

Usando exponenciación ingenua:

```python
%time naive_exp_mod(x, 100)
```

Como podemos observar, no hay una diferencia muy significativa entre ambos algoritmos por el momento. Elevaremos ahora el punto base a la potencia $n = 100000$. Con exponenciación binaria:

```python
%time exp_mod(x, 100000)
```

Usando exponenciación ingenua:

```python
%time naive_exp_mod(x, 100000)
```

Podemos notar que en este caso ya existe una diferencia significativa entre ambos algoritmos. Repetiremos ahora este proceso tomando como potencia $n = 10000000$. Con exponenciación binaria:

```python
%time exp_mod(x, 10000000)
```

Usando exponenciación ingenua:

```python
%time naive_exp_mod(x, 10000000)
```

Aquí podemos notar como la diferencia entre ambos algoritmos se va haciendo más palpable. Finalmente, tomaremos como potencia $n = 1000000000$. Con exponenciación binaria:

```python
%time exp_mod(x, 1000000000)
```

Usando exponenciación ingenua:

```python
%time naive_exp_mod(x, 1000000000)
```

Aquí ya poemos notar un efecto mucho más significativo. Queda claro que si tomamos potencias mucho mayores el cálculo por exponenciación ingenua se vuelve rápidamente imposible de realizar en un tiempo razonable.
