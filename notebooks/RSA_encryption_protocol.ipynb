Una vez que ya hemos cargado las librerías necesarias, empezamos por generar nuestras claves *pública* y *privada:*

```python
(public_key, private_key) = rsa_key_generation(2048)
```

Mostramos ahora la clave privada que generamos. En la práctica, es muy importante que esta información no se divulgue.

```python
print(f"Clave privada generada (no divulgar):\n")
print(f"Exponente público (e): {private_key[0]}\n")
print(f"Módulo generado (n): {private_key[1]}\n")
```

La clave pública, por otro lado, se debe de compartir con las partes interesadas, a fín de que puedan encriptar sus mensajes. En este caso la clave pública a compartir es:

```python
print(f"Clave pública generada: {public_key}\n")
```

Para demostrar como funciona este proceso, procedemos ahora a generar un mensaje que deseamos encriptar. Empezaremos con el simple 'Hello World!'

```python
plain_text = 'Hello World!'
```

Antes de poder encriptar este mensaje, necesitamos codificarlo en forma numérica:

```python
plain_message = encode(plain_text)
plain_message
```

Utilizando la clave pública, podemos ahora generar el *mensaje encriptado*

```python
encrypted_message = rsa_encryption(public_key, plain_message)
encrypted_message
```

Si este mensaje fuera interceptado por un adversario (digamos Eva) este sería el mensaje de texto que obtendría:

```python
encrypted_text = decode(encrypted_message)
encrypted_text
```

Por otro lado, Alicia al momento de recibir este mensaje podría desencriptarlo primero utilizando su clave privada:

```python
decrypted_message = rsa_decryption(private_key, encrypted_message)
decrypted_message
```

Para poder leer este mensaje basta con decodificarlo a mensaje de texto:

```python
decrypted_text = decode(decrypted_message)
decrypted_text
```
